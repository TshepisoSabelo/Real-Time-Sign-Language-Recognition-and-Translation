```python
images, labels = dl.load_dataset()
```

```python
# Example using Scikit-learn
from sklearn.model_selection import train_test_split

# Assume X contains your features and y contains your labels/targets
# First split into training+validation (85%) and test (15%)
X_train_val, X_test, y_train_val, y_test = train_test_split(
    images, labels, test_size = 0.15, random_state = 42, stratify = labels # Use stratify for classification
)

# Calculate the validation size relative to the train_val set
# If test is 15%, validation is 15%, then validation is 15 / (100-15) = 15/85 = ~0.176
val_size_relative = 0.15 / (1.0 - 0.15)

# Split train_val into final training and validation sets
X_train, X_val, y_train, y_val = train_test_split(
    X_train_val, y_train_val, test_size = val_size_relative, random_state = 42, stratify=y_train_val # Use stratify again
)

print(f"Original dataset size: {len(images)}")
print(f"Training set size: {len(X_train)}")
print(f"Validation set size: {len(X_val)}")
print(f"Test set size: {len(X_test)}")
```

```python
print(f"Images shape: {images.shape}")
print(f"X_train shape: {X_train.shape}")
```

```python
model = nw.Network()

model.add_layer(layers.Convolutional(input_shape = (1, 64, 64), kernel_size = 3, layer_size = 32))
model.add_layer(layers.Convolutional(input_shape = (1, 62, 62), kernel_size = 3, layer_size = 32))
model.add_layer(layers.Flatten()) #flattened output shape will be (8 * 26 * 26,) = (5408,1)
model.add_layer(layers.Dense((115200, 1), 1000, ReLu))
model.add_layer(layers.Dense((1000, 1), 264, ReLu))
model.add_layer(layers.Dense((264, 1), 128, ReLu))
model.add_layer(layers.Dense((128, 1), 28, Softmax))
#model.add_layer(layers.Dense(input_shape=(128, 1), output_size=64))
#model.add_layer(layers.Dense(input_shape=(128, 1), output_size=29))
#followed by softmax activation
#then the loss function
```

training

```python
model.train(loss = losses.categorical_cross_entropy,
            x_train = X_train,
            y_train = y_train,
            X_val = X_train_val,
            y_val = y_train_val,
            epochs = 25,
            learning_rate = 0.01,
            verbose = True)

#model.test(x_train=x_train, y_train=y_train, activation = activation.softmax, loss=losses.categorical_cross_entropy)
```
